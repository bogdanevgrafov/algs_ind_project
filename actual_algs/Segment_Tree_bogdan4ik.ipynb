:# Зачем нам вообще это

Пусть есть массив:

```text
[2, 5, 1, 4, 3, 7, 6, 8]
```

Нужно поддерживать две операции:

```text
SUM L R - получить сумму на [L..R]
SET i x - заменить a[i] на x
```

Обычный проход:

```text
SUM → O(n)
```

Prefix Sum:

```text
SUM → O(1)
SET → O(n)
```

Но теперь операции постоянно чередуются:

```text
SET
SUM
SET
SUM
...
```

Нам нужна структура, в которой **и запрос, и изменение работают быстро**

# Идея структуры

Мы заранее храним агрегаты для разных кусочков массива:

```text
                        [0..7] = 36
                 /                      \
          [0..3] = 12                [4..7] = 24
          /       \                   /       \
     [0..1]=7   [2..3]=5          [4..5]=10  [6..7]=14
       / \        / \              / \        / \
      2   5       1   4            3   7       6   8
```

Каждая вершина отвечает за свой отрезок

Для суммы значение внутренней вершины получается объединением двух детей:

```text
left + right
```

Ожидаемые сложности:

```text
build  → O(n)
query  → O(log n)
update → O(log n)
```

# Как дерево хранится в памяти

Отдельные объекты `Node` нам не нужны. Мы можем хранить дерево в обычном списке:

```python
tree = [0] * (4 * n)
```

Возьмем некоторое соглашение:

```text
v              - номер текущей вершины
2 * v          - левый ребенок
2 * v + 1      - правый ребенок
```

Кроме номера вершины рекурсивным методам нужно знать, **за какой отрезок она отвечает** (надеюсь это дял вас очевидно):

```text
tl - левая граница сегмента
tr - правая граница сегмента
```

Метод получает:

```text

```


# Что должен делать `query`

Внешний метод вызывается просто:

```python
st.query(l, r)
```

Его задача небольшая:

- проверить корректность границ
- обработать пустой массив, если вы хотите его поддерживать
- запустить внутренний `_query` **с корня дерева**

Сам поиск ответа происходит в `_query`

# Что должен делать `_query`

Текущая вершина отвечает за `[tl..tr]`, а нас интересует `[l..r]`

Для каждой вершины нужно определить один из трех случаев:

### 1. Текущий сегмент полностью снаружи запроса

Он ничего не добавляет к сумме

Подумайте, какое значение можно вернуть, чтобы оно **не изменило результат сложения**

### 2. Текущий сегмент полностью внутри запроса

В `tree[v]` уже лежит готовый ответ

Глубже идти не нужно

### 3. Сегменты пересекаются частично

Взять `tree[v]` целиком нельзя

Нужно обратиться к дочерним сегментам и объединить их результаты

# Что должен делать `update`

Внешний вызов:

```python
st.update(pos, value)
```

должен:

- проверить позицию
- при необходимости обновить сохранённый исходный массив
- запустить внутренний `_update` от корня

Само движение по дереву выполняет `_update`

# Что должен делать `_update`

Изменяется только один элемент массива.

Подумайте:

- В каком листе дерева находится позиция `pos`?
- Как на каждом уровне понять, идти в левого или правого ребенка?
- Что изменить, когда дошли до нужного листа?
- Какие вершины после этого стали неактуальными?
- Как пересчитать вершину после возвращения из рекурсии?
- Нужно ли вообще заходить в соседнюю ветку дерева?

Главная идея:

> Меняем один лист и пересчитываем только его предков до корня

# Теперь соберите весь класс

Не пытайтесь сразу написать все идеально. Удобный порядок:

```text
1. __init__ + _build
2. query + _query
3. update + _update
4. тесты
```

````python
class SegmentTree:
    def __init__(self, a):
        self.n = len(a)
        self.a = list(a)
        self.tree = [0] * (4 * max(self.n, 1))
        if self.n > 0:
            self._build(1, 0, self.n - 1)

    def _build(self, v, tl, tr):
        if tl == tr:
            self.tree[v] = self.a[tl]
            return
        tm = (tl + tr) // 2
        self._build(2 * v, tl, tm)
        self._build(2 * v + 1, tm + 1, tr)
        self.tree[v] = self.tree[2 * v] + self.tree[2 * v + 1]

    def query(self, l, r):
        if self.n == 0:
            raise ValueError("Массив пустой")
        if not (0 <= l <= r < self.n):
            raise ValueError(f"Неверные границы: {l}, {r}")
        return self._query(1, 0, self.n - 1, l, r)

    def _query(self, v, tl, tr, l, r):
        if r < tl or tr < l:          # сегмент снаружи запроса
            return 0
        if l <= tl and tr <= r:       # сегмент целиком внутри запроса
            return self.tree[v]
        tm = (tl + tr) // 2           # пересекаются частично
        return (self._query(2 * v, tl, tm, l, r)
                + self._query(2 * v + 1, tm + 1, tr, l, r))

    def update(self, pos, value):
        if not (0 <= pos < self.n):
            raise ValueError(f"Неверная позиция: {pos}")
        self.a[pos] = value
        self._update(1, 0, self.n - 1, pos, value)

    def _update(self, v, tl, tr, pos, value):
        if tl == tr:
            self.tree[v] = value
            return
        tm = (tl + tr) // 2
        if pos <= tm:
            self._update(2 * v, tl, tm, pos, value)
        else:
            self._update(2 * v + 1, tm + 1, tr, pos, value)
        self.tree[v] = self.tree[2 * v] + self.tree[2 * v + 1]
````

# Тест 1. Построение

Если `build` работает правильно, корень для этого массива должен хранить сумму всего массива

````python
a = [2, 5, 1, 4, 3, 7, 6, 8]
st = SegmentTree(a)

assert st.tree[1] == 36
print("BUILD")
````

# Тест 2. Запросы

````python
a = [2, 5, 1, 4, 3, 7, 6, 8]
st = SegmentTree(a)

tests = [
    (0, 0, 2),
    (0, 3, 12),
    (2, 6, 21),
    (4, 7, 24),
    (0, 7, 36),
    (7, 7, 8),
]

for l, r, expected in tests:
    actual = st.query(l, r)
    assert actual == expected, (
        f"query({l}, {r}) → {actual}, ожидалось {expected}"
    )

print("QUERY")
````

# Тест 3. Обновление

После:

# Тест 4. Несколько изменений подряд

Segment Tree нужен именно для сценария:

```text
update → query → update → query
```

````python
a = [2, 5, 1, 4, 3, 7, 6, 8]
st = SegmentTree(a)

assert st.query(2, 6) == 21

st.update(3, 10)
assert st.query(2, 6) == 27

st.update(5, 1)
assert st.query(2, 6) == 21

st.update(0, 100)
assert st.query(0, 1) == 105

print("УРА МЫ КРУТЫЕ")
````

# Интеграция в индивидуальный проект

Теперь примените Segment Tree к своему датасету

## Шаг 1. Выберите последовательность

Нужно выбрать:

- поле, задающее порядок записей
- числовое поле, которое потенциально может изменяться

Например:

```text
дата/время → порядок
количество посещений → значение
```

или:

```text
номер события → порядок
стоимость / нагрузка / рейтинг → значение
```

````python
import time
import random
import numpy as np
import pandas as pd

DATA_PATH = "https://data.cityofchicago.org/resource/ajtu-isnz.csv?$limit=200000&$where=trip_start_timestamp%3E%3D%272025-01-01T00:00:00%27"

df = pd.read_csv(DATA_PATH)
print("Размер:", df.shape)
````

Датасет: Chicago Taxi Trips (поездки такси в Чикаго, 200 000 строк).

**Порядок записей:** `trip_start_timestamp` (время начала поездки). Поездки сортируем по времени, номер поездки в этом порядке - позиция в дереве.

**Значение:** `fare` (тариф) в центах, целые числа. Пропуски заменяем на 0, строки не удаляем.

## Шаг 2. Сформулируйте динамический сценарий

Опишите две операции вашей системы:

```text
UPDATE i x
```

- изменить значение одной записи

```text
QUERY L R
```

- получить `SUM`, `MIN` или `MAX` на диапазоне

Сценарий должен иметь смысл именно для ваших данных

**UPDATE i x:** исправить тариф i-й поездки на x центов (например, после возврата или корректировки).

**QUERY L R:** получить суммарную выручку поездок с номера L по номер R (то есть за период времени).

Для аналитика это значит: выручку за любой период можно получить быстро, даже если тарифы постоянно исправляют.

## Шаг 3. Реализуйте индекс

Создайте отдельный класс или модуль, например:

```python
class SegmentTreeIndex:
    ...
```

В проекте должно быть видно:

```text
исходные данные
↓
построение Segment Tree
↓
update
↓
range query
```

````python
class SegmentTreeIndex:
    def __init__(self, df):
        df = df.copy()
        df["start"] = pd.to_datetime(df["trip_start_timestamp"])
        df = df.sort_values("start", kind="stable").reset_index(drop=True)

        self.times = df["start"].to_numpy()
        fares = np.round(df["fare"].fillna(0).to_numpy() * 100).astype(np.int64)

        self.tree = SegmentTree(fares.tolist())

    def update_fare(self, i, cents):
        self.tree.update(i, cents)

    def revenue(self, l, r):
        return self.tree.query(l, r)

    def revenue_between(self, t_from, t_to):
        # переводим период времени в диапазон номеров поездок
        l = int(np.searchsorted(self.times, np.datetime64(t_from), side="left"))
        r = int(np.searchsorted(self.times, np.datetime64(t_to), side="right")) - 1
        if l > r:
            return 0
        return self.tree.query(l, r)
````

````python
start = time.perf_counter()
index = SegmentTreeIndex(df)
build_ms = (time.perf_counter() - start) * 1000
print(f"Построение: {build_ms:.1f} ms, поездок в дереве: {index.tree.n}")

n = index.tree.n
print("Выручка по всем поездкам, центов:", index.revenue(0, n - 1))
print("Выручка поездок 100..200, центов:", index.revenue(100, 200))

old = index.tree.a[150]
index.update_fare(150, old + 1000)
print("После update (+1000 центов в поездке 150):", index.revenue(100, 200))

t_from = index.times[0]
t_to = index.times[n // 2]
print("Выручка за период первой половины данных, центов:", index.revenue_between(t_from, t_to))
````

## Шаг 4. Сравните с наивным решением

Создайте смешанную нагрузку:

```text
update
query
update
query
...
```

Например, не менее `1000` операций

Сравните:

```text
обычный проход по диапазону
vs
Segment Tree
```

Зафиксируйте:

```text
время построения
время обработки всех операций
количество update
количество query
```

````python
random.seed(1)
N_OPS = 2000      


ops = []
for k in range(N_OPS):
    if k % 2 == 0:
        ops.append(("update", random.randrange(n), random.randrange(0, 20000)))
    else:
        l = random.randrange(n)
        r = random.randrange(l, n)
        ops.append(("query", l, r))

n_update = sum(1 for op in ops if op[0] == "update")
n_query = sum(1 for op in ops if op[0] == "query")

#способ 1: Segment Tree
index = SegmentTreeIndex(df)       
tree_answers = []
start = time.perf_counter()
for op, x, y in ops:
    if op == "update":
        index.update_fare(x, y)
    else:
        tree_answers.append(index.revenue(x, y))
tree_ms = (time.perf_counter() - start) * 1000

#способ 2: обычный проход по диапазону 
naive = SegmentTreeIndex(df).tree.a[:]
naive_answers = []
start = time.perf_counter()
for op, x, y in ops:
    if op == "update":
        naive[x] = y
    else:
        naive_answers.append(sum(naive[x:y + 1]))
naive_ms = (time.perf_counter() - start) * 1000

assert tree_answers == naive_answers, "Ответы дерева и обычного прохода не совпали"

print("Количество update:", n_update)
print("Количество query:", n_query)
print(f"Время построения дерева: {build_ms:.1f} ms")
print(f"Segment Tree, все операции: {tree_ms:.1f} ms")
print(f"Обычный проход, все операции: {naive_ms:.1f} ms")
print(f"Ответы совпали, дерево быстрее в {naive_ms / tree_ms:.1f} раз")
````

### Итог

**Что получилось:**  
Ответы дерева и обычного прохода совпали на всех запросах.

**Почему так:**  
Обычный проход по диапазону стоит O(n) на каждый запрос, а Segment Tree - O(log n). Update в обычном списке стоит O(1), в дереве O(log n), но на смешанной нагрузке запросы важнее.

**Цена за скорость:**  
Время построения дерева и память (список из 4n значений).
